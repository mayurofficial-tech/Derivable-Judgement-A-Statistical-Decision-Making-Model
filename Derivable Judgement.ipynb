# Part A – Theoretical Foundation (Short Notes & Explanation)

---
1. What is inferential statistics?
- Inferential statistics is the branch of statistics used to draw conclusions about a population based on sample data. It helps in prediction, estimation, and decision-making.
---
2. What is Hypothesis Testing and its Components?
- Hypothesis testing is a statistical method used to determine whether a claim about a population is true. Its main components are Null Hypothesis (H₀), Alternative Hypothesis (H₁), Significance Level (α), Test Statistic, and P-value.
---
3. Explain Confidence Interval and Critical Value.
- A confidence interval is a range of values that is likely to contain the true population parameter with a certain confidence level (e.g., 95%). A critical value is the cutoff point used to decide whether to reject the null hypothesis.
---
4. Define P-value.
- The p-value is the probability of obtaining results as extreme as the observed data, assuming the null hypothesis is true. If the p-value is less than the significance level (α), the null hypothesis is rejected.
---
5. Differentiate Type I and Type II Errors.
- A Type I Error occurs when a true null hypothesis is incorrectly rejected (False Positive). A Type II Error occurs when a false null hypothesis is incorrectly accepted (False Negative).
---
6. Brief Description of Z-test, T-test, Chi-square Test, and ANOVA.
- A Z-test is used to compare means when the population standard deviation is known and the sample size is large. A T-test compares means when the population standard deviation is unknown. A Chi-square test checks the association between categorical variables, while ANOVA compares the means of three or more groups.
---
7. What is Covariance?
- Covariance measures the direction of the relationship between two variables. A positive covariance indicates variables move together, while a negative covariance indicates they move in opposite directions.
---
8. What is Correlation?
- Correlation measures the strength and direction of the relationship between two variables. Its value ranges from -1 to +1, where +1 indicates a perfect positive relationship and -1 indicates a perfect negative relationship.

# Part B – Data Analysis & Testing Tasks:

```python
# 1-Formulate at least two hypotheses from the dataset.
# HYPOTHESIS 1:smoking status vs diabetes
# H0: smaking status has no effect on diabetes prevalence
# H1: smoking status affects diabetes prevalence
#  USE - CHI-SQUARE Test

# HYPOTHESIS 2: Age Group vs BMI
# H0: The mean Bmi is the same across all age groups
# H1: At least one age group's mean Bmi is different
#  USE - ANOVA
```

```python
# 2-calculate Confidence interval for age,weight
# age
from scipy import stats
import numpy as np
import pandas as pd
DF=pd.read_csv("health_records_dataset.csv")
mean=DF['age'].mean()
std=DF['age'].std()
n=len(DF)

ci=stats.t.interval(
    confidence=0.95,
    df=n-1,
    loc=mean,
    scale=std/np.sqrt(n)
)
print("95% Ci for Age:",ci)

# weight
mean=DF['weight'].mean()
std=DF['weight'].std()
n=len(DF)

ci=stats.t.interval(
    confidence=0.95,
    df=n-1,
    loc=mean,
    scale=std/np.sqrt(n)
)

print("95% Ci for weight:",ci)
```

```python
# 3-Criticak value and p_value
# p-Test whether averahe Bmi is 25
# H0: u=25
# H1: u!=25

t_stat,p_value=stats.ttest_1samp(DF["bmi"],25)

print("T Statistic:",t_stat)

# p_value
print("P value:",p_value)

# critical t
al=0.05
df=len(DF["bmi"])-1
critical_t=stats.t.ppf(1-al/2,df)
print("Critical value:",critical_t)

# decision by p_value
if p_value<al:
    print("p_value decision is reject H0")
else:
    print("p_value decision is fail to reject H0")

# decision by critical value
if t_stat<-critical_t or t_stat>critical_t:
    print("Critical decision is reject H0")
else:
    print("Critical decision is fail to reject H0")
```

```python
# 4-independent Two-sample T_test
# p-check whether the mean Bmi of male and females is different
# H0: Mean Bmi of male = mean Bmi of female
# H1: mean bmi of male != mean bmi of female
male=DF[DF['gender']=="Male"]["bmi"]
female=DF[DF['gender']=='Female']['bmi']
al=0.05
stat,p=stats.levene(male,female )
if p<0.05:
    print("Equal Variances")
else:
    print("Unequal Variances")

values=stats.ttest_ind(male,female,equal_var=False)
t_stats=values.statistic
df=values.df
p_value=values.pvalue
critical_t=stats.t.ppf(1-al/2,df)

# decision
if t_stats>critical_t or t_stats<-critical_t:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

```python
# 5-Chi-square Test)Smoking Status vs Diabetes
# p-check whether smoking status and diabetes are associated
# H0: smaking Status and diabetes are independent(No association)
# H1: smoking status and diabetes are depandent(association)

table=pd.crosstab(
    DF["smoking_status"],
    DF["diabetes"]
)

chi2_stat,p_value,df,expected=stats.chi2_contingency(table)
print("Chi square:",chi2_stat)
print("p_value:",p_value)
print('Degrres if Freedom:',df) 
# critical value
critical_value=stats.chi2.ppf(1-al,df)
print("critical value:",critical_value)

# decision

if chi2_stat>critical_value:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

```python
# 6-AVONA
# p-check whether the mean BMI differs acriss age groups
# H0: Mean BMI is the same for all age groups
# H1: At least one age group"s mmean BMI is defferent
g1=DF[DF["age_group"]=='18-25']["bmi"]
g2=DF[DF["age_group"]=='36-45']["bmi"]
g3=DF[DF["age_group"]=='60+']["bmi"]
g4=DF[DF["age_group"]=='46-60']["bmi"]
g5=DF[DF["age_group"]=='26-35']["bmi"]
f_stat,p_value=stats.f_oneway(g1,g2,g3,g4,g5)

print("F_Statistic:",f_stat)
print("P_value:",p_value)

# decision
if p_value<0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

```python
# 7-Covariance and correlation(Age vs BMI)
# Covariance
cov=np.cov(
    DF["age"],
    DF["bmi"]
)[0,1]
print("Covariance:",cov)
if cov>0:
    print("Age and Bmi tend to increase together")
else:
    print("Age are increases,BMI tends to decrease")

# correlation
corr=DF["age"].corr(DF["bmi"])
print("Correlation:",corr)
print("weak Positive")
```
